## 📁 ตั้งค่าโฟลเดอร์สำหรับเก็บไฟล์

```python
import os
from pathlib import Path

# กำหนดโฟลเดอร์สำหรับเก็บไฟล์ MP3
OUTPUT_DIR = Path("./downloads")
OUTPUT_DIR.mkdir(exist_ok=True)

print(f"📁 ไฟล์จะถูกเก็บไว้ที่: {OUTPUT_DIR.absolute()}")
```

## 🎵 ดาวน์โหลด MP3 จาก URL เดียว

```python
import yt_dlp

def download_mp3(url: str, output_dir: Path = OUTPUT_DIR):
    """
    ดาวน์โหลด MP3 จาก YouTube URL
    
    Args:
        url: YouTube URL
        output_dir: โฟลเดอร์สำหรับเก็บไฟล์
    """
    ydl_opts = {
        'format': 'bestaudio/best',
        'postprocessors': [{
            'key': 'FFmpegExtractAudio',
            'preferredcodec': 'mp3',
            'preferredquality': '320',
        }],
        'outtmpl': str(output_dir / '%(title)s.%(ext)s'),
        'quiet': False,
        'no_warnings': False,
    }
    
    try:
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            info = ydl.extract_info(url, download=True)
            print(f"\n✅ ดาวน์โหลดสำเร็จ: {info.get('title', 'Unknown')}")
            return True
    except Exception as e:
        print(f"❌ เกิดข้อผิดพลาด: {e}")
        return False
```

```python
# 🎯 ใส่ URL ที่ต้องการดาวน์โหลดที่นี่
youtube_url = "https://www.youtube.com/watch?v=YOUR_VIDEO_ID"

download_mp3(youtube_url)
```

## 📋 ดาวน์โหลดหลาย URL พร้อมกัน

```python
def download_multiple_mp3(urls: list, output_dir: Path = OUTPUT_DIR):
    """
    ดาวน์โหลด MP3 จากหลาย YouTube URLs
    
    Args:
        urls: รายการ YouTube URLs
        output_dir: โฟลเดอร์สำหรับเก็บไฟล์
    """
    success_count = 0
    fail_count = 0
    
    for i, url in enumerate(urls, 1):
        print(f"\n{'='*50}")
        print(f"📥 กำลังดาวน์โหลด ({i}/{len(urls)}): {url}")
        print(f"{'='*50}")
        
        if download_mp3(url, output_dir):
            success_count += 1
        else:
            fail_count += 1
    
    print(f"\n{'='*50}")
    print(f"📊 สรุป: สำเร็จ {success_count} | ล้มเหลว {fail_count}")
    print(f"{'='*50}")
```

```python
# 🎯 ใส่รายการ URLs ที่ต้องการดาวน์โหลด
urls_to_download = [
    "https://www.youtube.com/watch?v=VIDEO_ID_1",
    "https://www.youtube.com/watch?v=VIDEO_ID_2",
    "https://www.youtube.com/watch?v=VIDEO_ID_3",
]

download_multiple_mp3(urls_to_download)
```

## 🎶 ดาวน์โหลดทั้ง Playlist

```python
def download_playlist(playlist_url: str, output_dir: Path = OUTPUT_DIR):
    """
    ดาวน์โหลดทั้ง Playlist เป็น MP3
    
    Args:
        playlist_url: YouTube Playlist URL
        output_dir: โฟลเดอร์สำหรับเก็บไฟล์
    """
    ydl_opts = {
        'format': 'bestaudio/best',
        'postprocessors': [{
            'key': 'FFmpegExtractAudio',
            'preferredcodec': 'mp3',
            'preferredquality': '320',
        }],
        'outtmpl': str(output_dir / '%(playlist_title)s/%(playlist_index)s - %(title)s.%(ext)s'),
        'quiet': False,
        'no_warnings': False,
        'ignoreerrors': True,  # ข้ามวิดีโอที่มีปัญหา
    }
    
    try:
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            ydl.download([playlist_url])
            print(f"\n✅ ดาวน์โหลด Playlist สำเร็จ!")
            return True
    except Exception as e:
        print(f"❌ เกิดข้อผิดพลาด: {e}")
        return False
```

```python
# 🎯 ใส่ Playlist URL ที่นี่
playlist_url = "https://www.youtube.com/playlist?list=PLbzGR7H3FyUQXT-XBRNXwVdWB3FDeQBwZ"
download_playlist(playlist_url)
```

## 📂 ดูไฟล์ที่ดาวน์โหลดแล้ว

```python
import os

def list_downloaded_files(directory: Path = OUTPUT_DIR):
    """แสดงรายการไฟล์ที่ดาวน์โหลดแล้ว"""
    if not directory.exists():
        print("📁 ยังไม่มีไฟล์ที่ดาวน์โหลด")
        return
    
    files = list(directory.rglob("*.mp3"))
    
    if not files:
        print("📁 ยังไม่มีไฟล์ MP3 ที่ดาวน์โหลด")
        return
    
    print(f"🎵 ไฟล์ MP3 ที่ดาวน์โหลดแล้ว ({len(files)} ไฟล์):\n")
    
    for file in sorted(files):
        size_mb = file.stat().st_size / (1024 * 1024)
        print(f"  📄 {file.name} ({size_mb:.2f} MB)")

list_downloaded_files()
```
